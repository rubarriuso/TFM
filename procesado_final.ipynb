```python
import xarray as xr
import numpy as np
```

```python
#Variables fisicas
phy = xr.open_zarr("./resources/copernicus_marine_service/regridded/obs-mob_glo_phy_regridded.zarr")
to_surface = phy["to"].sel(depth=0, method="nearest").drop_vars("depth") #temperature
so_surface = phy["so"].sel(depth=0, method="nearest").drop_vars("depth") #salinity
ugo_surface = phy["ugo"].sel(depth=0, method="nearest").drop_vars("depth") #u geostrofic
vgo_surface = phy["vgo"].sel(depth=0, method="nearest").drop_vars("depth") #v geostrofic
mlotst = phy["mlotst"] #mixed layer depth
zo_surface = phy["zo"].sel(depth=0, method="nearest").drop_vars("depth") #geostrophic geopotential height
```

```python
#Save
to_surface.to_netcdf("./data/processed/dynamic/to_surface.nc")
so_surface.to_netcdf("./data/processed/dynamic/so_surface.nc")
ugo_surface.to_netcdf("./data/processed/dynamic/ugo_surface.nc")
vgo_surface.to_netcdf("./data/processed/dynamic/vgo_surface.nc")
mlotst.to_netcdf("./data/processed/dynamic/mixed_layer.nc")
zo_surface.to_netcdf("./data/processed/dynamic/zo_surface.nc")
```

```python
#Variables biogeoquimicas
cdm_ds = xr.open_zarr("./resources/copernicus_marine_service/regridded/obs-oc_glo_optics.zarr")
chl_ds = xr.open_zarr("./resources/copernicus_marine_service/regridded/obs-oc_glo_plankton.zarr")
pp_ds = xr.open_zarr("./resources/copernicus_marine_service/regridded/obs-oc_glo_pp.zarr")
transp_ds = xr.open_zarr("./resources/copernicus_marine_service/regridded/obs-oc_glo_transp.zarr")
spm = transp_ds["SPM"] #Concentration of suspender mater
zsd = transp_ds["ZSD"] #Secchi depth of water
chl = chl_ds["CHL"] #chlorophyl
pp = pp_ds["PP"] #primary productivity
cdm = cdm_ds["CDM"] #absorbtion coefficent
```

```python
#Save
spm.to_netcdf("./data/processed/dynamic/spm.nc")
zsd.to_netcdf("./data/processed/dynamic/zsd.nc")
chl.to_netcdf("./data/processed/dynamic/chl.nc")
pp.to_netcdf("./data/processed/dynamic/pp.nc")
cdm.to_netcdf("./data/processed/dynamic/cdm.nc")
```

```python
#Batimetría
baty = xr.open_zarr("./resources/GEBCO/GEBCO_regridded.zarr")
baty["elevation"] = baty["elevation"].where(
    baty["elevation"] <= 0, np.nan) #put nan to values greater than 0
baty["elevation"] = np.abs(baty["elevation"]) #change depth to positive
baty = baty.rename({"elevation": "depth"}) #rename elevation to depth
depth = baty["depth"]
```

```python
depth.to_netcdf("./data/processed/static/depth.nc")
```
