```python
import xarray as xr
import geopandas as gpd
import matplotlib.pyplot as plt
import geopandas as gpd
import cartopy.crs as ccrs
import cartopy.feature as cfeature
import pandas as pd
import imageio
import io
```

```python
temp_bottom = xr.open_dataset("./data/processed/dynamic/temp_bottom.nc")
so_surface = xr.open_dataset("./data/processed/dynamic/so_surface.nc")
to_surface = xr.open_dataset("./data/processed/dynamic/to_surface.nc")
mixed = xr.open_dataset("./data/processed/dynamic/mixed_layer.nc")
pp = xr.open_dataset("./data/processed/dynamic/pp.nc")
```

```python
eez_path = "./resources/Marine_Regions/World_EEZ_v12/eez_v12.shp"
min_lon, min_lat, max_lon, max_lat = [-69.61, -60., -50., -32.45] #SW Atlantic bbox (roundeed to the second decimal)
bbox = [min_lon, min_lat, max_lon, max_lat]
eez_gdf = gpd.read_file(eez_path, bbox=tuple(bbox))
```

```python
def plot_env(ds, month, env, leglabel, title, vmin, vmax, colormap):
    fig = plt.figure(figsize=(20, 12))
    ax = plt.axes(projection=ccrs.PlateCarree())

    ax.set_extent([
        bbox[0] + 2, # Left
        bbox[2] - 2, # Right
        bbox[1] + 3, # Bottom
        bbox[3] - 5  # Top
    ], crs=ccrs.PlateCarree())


    plot_data = ds[env].sel(time=month, method="nearest")

    plot_data.plot(ax=ax, transform=ccrs.PlateCarree(),
                        vmin = vmin, vmax = vmax,
                        cmap=colormap,
                        add_colorbar=True, cbar_kwargs={'label': leglabel, 'extend': 'both'},
                        zorder=1)

    ax.add_feature(cfeature.LAND, facecolor='gray', edgecolor='gray', zorder=3)
    ax.add_geometries(eez_gdf.geometry, crs=ccrs.PlateCarree(), edgecolor='grey', facecolor='none')
    ax.add_feature(cfeature.BORDERS, linewidth=0.5, zorder=4)

    plt.title(f"{title} {month}", fontsize=14, fontweight='bold', pad=20)

    return fig
```

```python

months = pd.date_range(start="2013-01", end="2024-12", freq="MS").strftime("%Y-%m").tolist()

frames = []

for month in months:
    print(f"Processing: {month}") # To track progress
    
    # Generate the plot using your existing logic
    fig = plot_env(temp_bottom, month, "to", "Temperatura (ºC)", "Temperatura en el fondo", vmax=15, vmin=0, colormap="plasma") 
    
    # Convert plot to image frame
    buf = io.BytesIO()
    plt.savefig(buf, format='png', dpi=100, bbox_inches='tight')
    plt.close(fig)
    buf.seek(0)
    frames.append(imageio.v2.imread(buf))

# Save the final animation
imageio.mimsave(f'./data/gifs/bottom_temperature.gif', frames, fps=1)
print("Saved succesfully")
```
