GFW data raster 0.1x0.1 degree resolution from SWF bbox

```python
from models.GFW_auth import GFW_auth
import json
import requests
```

```python
with open("./data/bbox.json", "r") as f:
    bbox = json.load(f)

min_lon, min_lat, max_lon, max_lat = [
    bbox["min_lon"],
    bbox["min_lat"],
    bbox["max_lon"],
    bbox["max_lat"]
]
print(min_lon, min_lat, max_lon, max_lat)
```

```python
def get_fishing_effort_report(year, min_lon, min_lat, max_lon, max_lat):
    #To do: poner parametros a esta funcion de: date range, temporal resolution, arte de pesca...
    url="https://gateway.api.globalfishingwatch.org/v3/4wings/report"

    headers=GFW_auth().headers
    payload = {
        "geojson": {
            "type": "Polygon",
            "coordinates": [[
                [min_lon, min_lat],
                [max_lon, min_lat],
                [max_lon, max_lat],
                [min_lon, max_lat],
                [min_lon, min_lat]  # close the polygon
            ]]
        }
    }
    
    params = {
        'date-range': f"{year}-01-01,{year}-12-31",
        "spatial-resolution": "LOW",
        "temporal-resolution": "MONTHLY",
        "datasets[0]": "public-global-fishing-effort:latest",
        "format": "JSON",
        "group-by": "VESSEL_ID",
    }
    response = requests.post(url, params=params, headers=headers, json=payload)

    if response.status_code == 200:
        print("Response Status 200")
        data = response.json()
        return data
    else:
        print("Error:", response.status_code)
        print("Response text:", response.text)
```

```python
list_years = range(2012, 2025+1)
for year in list_years:
    print(f"Calling api for year {year}:")
    data = get_fishing_effort_report(year, min_lon, min_lat, max_lon, max_lat)

    with open(f"./resources/GFW/GFW_unprocessed/GFW_SWA_AIS_{year}.json", "w") as f:
        json.dump(data, f, indent=2)
```
